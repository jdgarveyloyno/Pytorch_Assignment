Part 7 — NumPy ↔ PyTorch

Create the following NumPy array:
```python

              [3., 4.]])
```

````python
# TODO: Convert the NumPy array into a PyTorch tensor.
a = np.array([[1., 2.],
              [3., 4.]])
A_tensor = torch.from_numpy(a)
````

````python
# TODO: print tensor A, shape and data type
print("Tensor A:\n", A_tensor)
print("Shape:", A_tensor.shape)
print("Data type:", A_tensor.dtype)
````

````python
# TODO: Add 10 to the A tensor
A_tensor_plus_10 = A_tensor + 10
````

````python
# TODO: Convert the resulting tensor back into a NumPy array.
a_array = A_tensor_plus_10.numpy()
````

````python
# TODO: Verify that the resulting object is a NumPy array.
print("Result:\n", a_array)
print("Is NumPy array?:", isinstance(a_array, np.ndarray))
````

Part 9 — Automatic Differentiation

PyTorch provides automatic differentiation using autograd.

````python
x = torch.tensor([[1., 2.],
                  [3., 4.]], requires_grad=True)

# TODO: Calculate y = 3x² + 2
y = 3 * (x ** 2) + 2
````

````python
# Then calculate:
z = sum(y)

# Use:
z.backward()
# to calculate: dz/dx

# Print:

print(y)
print(z)
print(x.grad)
````

* What does backward() do?
